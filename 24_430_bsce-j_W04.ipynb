```python
STUDENT_ID="24/430/bsce-j"
COHORT="CML3112"
import csv
with open("site_inspection_log.csv",newline="") as f:
    rows=list(csv.DictReader(f))
```

## valid_count (5 marks)
Store an integer count of rows whose crack_width_mm field is nonblank. Skip blank readings before numeric conversion.


```python
# Slide 59: skip blanks BEFORE converting text to numbers
values = []
for row in rows:
    text = row["crack_width_mm"]
    if text != "":
        values.append(float(text))

valid_count = len(values)
print(valid_count)
```

## mean_reading (5 marks)
Store the arithmetic mean of the nonblank crack_width_mm values as a number. Add the valid readings and divide by their count. Keep full precision.


```python
# Slides 18-20: running total, then total / count
total = 0
for reading in values:
    total = total + reading

mean_reading = total / valid_count
print(mean_reading)
```

## above_limit (5 marks)
Store an integer count of valid crack_width_mm readings strictly greater than 0.3. A reading equal to the threshold is not counted.


```python
# Slide 14: a counter. "Strictly greater" means > (not >=)
above_limit = 0
for reading in values:
    if reading > 0.3:
        above_limit = above_limit + 1

print(above_limit)
```

## group_means (5 marks)
Store a dictionary whose keys are the structure_id labels and whose values are the mean valid crack_width_mm readings within each group. Divide each group total by its own valid count. Keep full precision.


```python
# Slides 62-65: one running total and one count per group
totals = {}
counts = {}
for row in rows:
    text = row["crack_width_mm"]
    if text == "":
        continue                      # blank reading: skip (same policy as above)
    group = row["structure_id"]
    if group not in totals:
        totals[group] = 0
    totals[group] = totals[group] + float(text)
    counts[group] = counts.get(group, 0) + 1

group_means = {}
for group, total in totals.items():
    group_means[group] = total / counts[group]

print(group_means)
```
